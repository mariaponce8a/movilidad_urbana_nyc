# 02. Análisis Espacial y Temporal
Búsqueda de patrones espaciales y temporales.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
viajes = pd.read_parquet("../datos/procesados/viajes_procesados.parquet")
```

## Patrones Temporales


```python
# Demanda por hora del día
plt.figure(figsize=(10,5))
sns.countplot(data=viajes, x="hora_dia", palette="viridis")
plt.title("Demanda por Hora del Día")
plt.show()
```

```python
# Demanda por día de la semana (0=Lunes, 6=Domingo)
plt.figure(figsize=(10,5))
sns.countplot(data=viajes, x="dia_semana", palette="viridis")
plt.title("Demanda por Día de la Semana")
plt.show()
```

## Patrones Espaciales (Hot Spots)


```python
# Top 10 Zonas de Origen
top_origen = viajes["PU_Zone"].value_counts().head(10)
top_origen.plot(kind="bar", title="Top 10 Zonas de Origen (PickUp)", figsize=(10,5))
plt.show()
```
