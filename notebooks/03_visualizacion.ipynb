# 03. Visualizaciones y Mapas Interactivos
Usaremos folium y plotly para crear mapas interactivos sin dependencias externas a APIs.


```python
import pandas as pd
import geopandas as gpd
import folium
import plotly.express as px
```

```python
demanda = pd.read_csv("../datos/procesados/demanda_por_zona.csv")
```

```python
# Para pintar un mapa necesitamos los límites geográficos reales de las zonas.
# Descargar los shapefiles de TLC Taxi Zones: https://data.cityofnewyork.us/api/geospatial/d3c5-ddgc?method=export&format=GeoJSON
# En caso de no tenerlo, Folium permite visualizar puntos. Para este caso simularemos centroides o usaremos Plotly.
```

```python
# Gráfico interactivo temporal con Plotly
fig = px.bar(demanda.groupby("hora_dia")["total_viajes"].sum().reset_index(), x="hora_dia", y="total_viajes", title="Demanda Agregada por Hora (Interactivo)")
fig.show()
```
