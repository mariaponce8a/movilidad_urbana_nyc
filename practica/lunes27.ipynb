```python
class Persona:
    def __init__(self, nombre, edad, campus):
        self.nombre = nombre,
        self.edad = edad,
        self.campus = campus

    def __str__(self):
        return f"${self.nombre}, ${self.edad} anios, campus ${self.campus}"
    
class Estudiante(Persona): 
    def __init__(self, nombre, edad, campus, carrera):
        super().__init__(nombre, edad,campus),
        self.carrera = carrera,
        self.__calificaciones = []  

    def agregar_calificacion(self, nota):
        self.__calificaciones.append(nota)

    @property
    def promedio(self):
        if len(self.__calificaciones) == 0:
            return 0
        return sum(self.__calificaciones)/len(self.__calificaciones)

    def descripcion(self):
        return f"{self.nombre} — {self.carrera} — Promedio: {self.promedio:.2f}"
    


# Subclase Profesor que hereda de Persona
class Profesor(Persona):
    def __init__(self, nombre, edad, campus, facultad):
        super().__init__(nombre, edad, campus)
        self.facultad = facultad

    def asignar_calificacion(self, estudiante, calificacion):
        if isinstance(estudiante,Estudiante):
            estudiante.agregar_calificacion(calificacion)
        else:
            raise TypeError("El objeto no es un estudiante")

    def descripcion(self):
        return f"Prof. {self.nombre} — Facultad de {self.facultad}"


        
```

```python
#PRUEBAS
# Prueba tu implementación
estudiante = Estudiante("3", 20, "Norte", "Ingeniería de Software")
profesor = Profesor("Dr. López", 45, "Norte", "Ciencias de la Computación")

# El profesor asigna calificaciones
profesor.asignar_calificacion(estudiante, 85)
profesor.asignar_calificacion(estudiante, 90)
profesor.asignar_calificacion(estudiante, 58)


print(estudiante.descripcion())  # Ana — Ingeniería de Software — Promedio: 84.33
print(profesor.descripcion())    # Prof. Dr. López — Facultad de Ciencias de la Computación

# Validaciones
assert abs(estudiante.promedio - 77.66) < 0.01, "El promedio no es correcto"
print("\n✓ Ejercicio 1 completado")
```
